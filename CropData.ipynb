```python
for crop in ["Rice", "Wheat", "Maize"]:
    arr = df[crop].dropna().to_numpy()
    print(f"\n{crop} → Mean: {np.mean(arr):.2f}, Median: {np.median(arr):.2f}, Std Deviation: {np.std(arr):.2f}")
```

```python
print("\nTop 5 States (Rice):\n", df.groupby("State Name")["Rice"].sum().sort_values(ascending=False).head(5))
```

```python
# Top 5 Districts by Wheat Production
print("Top 5 Districts (Wheat):\n", df.groupby("Dist Name")["Wheat"].sum().sort_values(ascending=False).head(5))
```

```python
# Year with Max Production for Each Crop
for crop in ["Rice","Wheat","Maize"]:
    yearly = df.groupby("Year")[crop].sum()
    print(f"{crop} → Max Year: {yearly.idxmax()} ({yearly.max()})")
```

```python
# Year-on-Year Growth (%) for Rice
rice_growth = df.groupby("Year")["Rice"].sum().pct_change()*100
print("Rice Growth (%):\n", rice_growth)
```

```python
#Correlation Between Crops
print("Correlation Matrix:\n", df[["Rice","Wheat","Maize"]].corr())
```

```python
# State wise Average for All Crops
print("State wise Mean Production:\n", df.groupby("State Name")[["Rice","Wheat","Maize"]].mean().head())
```

```python
# Outlier Detection using IQR (Interquartile Range)
for crop in ["Rice","Wheat","Maize"]:
    arr = df[crop].dropna().to_numpy()
    q1, q3 = np.percentile(arr, [25, 75])
    iqr = q3 - q1
    outliers = arr[(arr < q1-1.5*iqr) | (arr > q3+1.5*iqr)]
    print(f"{crop} Outliers → {len(outliers)}")
```

```python
# District wise Highest Production Each Year of rice
rice_top_districts = df.groupby(["Year","Dist Name"])["Rice"].sum().reset_index()
print("\nTop Rice Districts Each Year:\n", rice_top_districts.sort_values(["Year","Rice"], ascending=[True,False]).groupby("Year").head(1))
```

```python
# Ratio of Rice to Wheat Production per Year
ratio = df.groupby("Year").agg({"Rice":"sum","Wheat":"sum"})
ratio["Rice/Wheat"] = ratio["Rice"]/ratio["Wheat"]
print("Rice/Wheat Ratio by Year:\n", ratio)
```

```python
# Identify Years with Zero Production
zero_years = df.groupby("Year")[["Rice","Wheat","Maize"]].sum()
print("\nYears with Zero Production:\n", zero_years[(zero_years==0).any(axis=1)])
```

```python
# Variance for Crops
for crop in ["Rice","Wheat","Maize"]:
    print(f"{crop} Variance:", np.var(df[crop].dropna()))
```

# Visualization

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette("viridis")
```

```python
crop_cols = [col for col in df.columns if 'PRODUCTION' in col.upper()]
total_crops = df[crop_cols].sum().sort_values(ascending=False)

plt.figure(figsize=(10,5))
sns.barplot(x=total_crops.index, y=total_crops.values, color='steelblue')
plt.xticks(rotation=75)
plt.title("Total Production by Crop")
plt.ylabel("Production (1000 tons)")
plt.show()
```

```python
# year wise
df['Total_Production'] = df[crop_cols].sum(axis=1)
yearly_prod = df.groupby('Year')['Total_Production'].sum().reset_index()

plt.figure(figsize=(8,4))
sns.lineplot(x='Year', y='Total_Production', data=yearly_prod, marker='o')
plt.title("Year-wise Total Crop Production Trend")
plt.show()
```

```python
# state wise average
state_avg = df.groupby('State Name')['Total_Production'].mean().sort_values(ascending=False).head(10)
plt.figure(figsize=(10,5))
sns.barplot(x=state_avg.values, y=state_avg.index)
plt.title("Top 10 States by Average Production")
plt.xlabel("Average Production (1000 tons)")
plt.show()
```

```python
# Cleaning and Correlation Heatmap
crop_cols = [
    'Rice',
    'Wheat',
    'Maize',
    'COTTON PRODUCTION (1000 tons)',
    'SUGARCANE PRODUCTION (1000 tons)'
]

clean_df = df[crop_cols].apply(pd.to_numeric, errors='coerce')
clean_df = clean_df.dropna()

plt.figure(figsize=(9,6))
sns.heatmap(clean_df.corr(), annot=True, cmap="coolwarm", vmin=-1, vmax=1, linewidths=0.4)
plt.title("Cleaned Correlation Between Crops", fontsize=15, color ="#2C5875")
plt.show()
```

```python
# distribution of crop production
plt.figure(figsize=(12,6))
for crop in crop_cols[:6]:
    sns.kdeplot(np.log1p(df[crop].dropna()), label=crop, fill=True)
plt.title("Log-Scaled Distribution of Crop Productions", fontsize=14)
plt.xlabel("Log(Production + 1)")
plt.ylabel("Density")
plt.legend()
plt.show()
```

```python
# year wise crop comparison
year_crop = df.groupby('Year')[crop_cols].sum().reset_index()
year_crop_melt = year_crop.melt(id_vars='Year', var_name='Crop', value_name='Production')

plt.figure(figsize=(12,6))
sns.lineplot(x='Year', y='Production', hue='Crop', data=year_crop_melt)
plt.title("Year-wise Crop Comparison")
plt.show()
```

```python
# pair plot
sns.pairplot(df[crop_cols[:4]])
plt.suptitle("Pairwise Relationship Between Crops", y=1.02)
plt.show()
```

```python
top_states = (
    df.groupby(['Year', 'State Name'])['Total_Production']
    .sum()
    .reset_index()
    .sort_values(['Year', 'Total_Production'], ascending=[True, False])
)

for year in df['Year'].unique()[:5]:
    data = top_states[top_states['Year'] == year].head(5)
    plt.figure(figsize=(8,4))
    sns.barplot(x='Total_Production', y='State Name', data=data, color='darkcyan')
    plt.title(f"Top 5 States in {year}")
    plt.show()
```

```python
plt.figure(figsize=(6,5))
sns.scatterplot(x='Rice', y='Wheat', data=df)
plt.title("Rice vs Wheat Production", color='darkred')
plt.show()
```

```python
# histograms
df[crop_cols].hist(figsize=(15,10), bins=30, color='yellowgreen')
plt.suptitle("Crop Production Distribution Histograms")
plt.show()
```

```python
avg_crop_state = df.groupby('State Name')[crop_cols].mean().head(10)
avg_crop_state.plot(kind='bar', stacked=True, figsize=(14,6))
plt.title("Average Crop Production by State")
plt.ylabel("Production (1000 tons)")
plt.show()
```

```python
crop_variances = df[crop_cols].var().sort_values(ascending=False)
plt.figure(figsize=(10,5))
sns.barplot(x=crop_variances.index, y=crop_variances.values, color='mediumaquamarine')
plt.xticks(rotation=75)
plt.title("Variance of Crop Production (Volatility)")
plt.show()
```

```python
# Average Production per Crop per Year
avg_crop_year = df.groupby('Year')[crop_cols].mean().reset_index()
avg_crop_year.plot(x='Year', figsize=(12,6))
plt.title("Average Crop Production Per Year")
plt.ylabel("Production (1000 tons)")
plt.show()
```

```python
pivot = df.pivot_table(index='State Name', values=crop_cols, aggfunc='sum')
plt.figure(figsize=(12,10))
sns.heatmap(pivot, cmap="YlGnBu")
plt.title("State-wise Crop Production Heatmap")
plt.show()
```

```python
# Use year_crop DataFrame and set "Year" as index for plotting
combined = year_crop.set_index("Year")

fig, ax = plt.subplots(figsize=(12, 6))
for col in combined.columns:
    ax.plot(
        combined.index,
        combined[col],
        linewidth=3,
        marker='D',
        markersize=6,
        label=col
    )
ax.set_title("Crop Production Trends in India (1970–2018)", fontsize=16, fontweight="bold", color="#2F4F4F")
ax.set_xlabel("Year", fontsize=12, labelpad=10)
ax.set_ylabel("Production (1000 tons)", fontsize=12, labelpad=10)
ax.legend(title="Crop Type", fontsize=10, title_fontsize=11, loc="upper left", frameon=True)

ax.grid(True, which='both', linestyle='--', alpha=0.6)
for spine in ax.spines.values():
    spine.set_visible(False)

plt.tight_layout()
plt.show()
```
